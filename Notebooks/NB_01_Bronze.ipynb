```python
source_file = "customers.csv"
bronze_table = "bronze_customers"
run_id = "manual"
```

```python
df = spark.read.csv(
    "Files/raw/" + source_file,
    header=True,
    inferSchema=False
)

print("CSV loaded successfully")
print("Rows:", df.count())
```

```python
from pyspark.sql import functions as F

df = df.withColumn("_source_file", F.lit(source_file))
df = df.withColumn("_load_ts", F.current_timestamp())
df = df.withColumn("_run_id", F.lit(run_id))
```

```python
df.write.format("delta").mode("overwrite").saveAsTable(bronze_table)
 
print("Created table:", bronze_table)
```

```python
spark.read.table(bronze_table).printSchema()
print("Rows:", spark.read.table(bronze_table).count())
```
